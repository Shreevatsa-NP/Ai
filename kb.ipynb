```python
from itertools import product

# Generate all possible truth values of P, Q, R
print("PROPOSITIONAL LOGIC - KNOWLEDGE BASE")
print("=" * 75)

print("\nKnowledge Base:")
print("1. Q -> P")
print("2. P -> ~Q")
print("3. Q OR R")

# Function to evaluate implication
def implies(a, b):
    return (not a) or b

# Truth table
print("\nTRUTH TABLE")
print("-" * 75)

print(f"{'P':^5} {'Q':^5} {'R':^5} {'Q->P':^8} {'P->~Q':^8} {'Q OR R':^8} {'KB':^8}")

models = []

for P, Q, R in product([True, False], repeat=3):

    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R

    KB = s1 and s2 and s3

    print(f"{str(P):^5} {str(Q):^5} {str(R):^5} "
          f"{str(s1):^8} {str(s2):^8} "
          f"{str(s3):^8} {str(KB):^8}")

    if KB:
        models.append((P, Q, R))

print("\n" + "=" * 75)
print("i) MODELS WHERE KB IS TRUE")
print("=" * 75)

for model in models:
    print("P =", model[0], " Q =", model[1], " R =", model[2])

def entails(conclusion):
    for P, Q, R in models:
        if not conclusion(P, Q, R):
            return False
    return True

print("\nii) Does KB entail R?")
print("Answer:", entails(lambda P, Q, R: R))

print("\niii) Does KB entail R -> P?")
print("Answer:", entails(lambda P, Q, R: implies(R, P)))

print("\niv) Does KB entail Q -> R?")
print("Answer:", entails(lambda P, Q, R: implies(Q, R)))

print("\n" + "=" * 75)
print("FINAL ANSWERS")
print("=" * 75)

print("i) KB Models:", len(models))
print("ii) KB entails R:", entails(lambda P, Q, R: R))
print("iii) KB entails R -> P:", entails(lambda P, Q, R: implies(R, P)))
print("iv) KB entails Q -> R:", entails(lambda P, Q, R: implies(Q, R)))
```
