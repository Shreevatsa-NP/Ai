```python

from itertools import combinations

# Required initial state
state = [3, 1, 2, 0]


def calculate_cost(state):
    cost = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def print_matrix(state):
    n = len(state)

    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print(" Q ", end="")
            else:
                print(" - ", end="")
        print()


def generate_neighbours(state):
    neighbours = []

    for i, j in combinations(range(len(state)), 2):
        new_state = state.copy()
        new_state[i], new_state[j] = new_state[j], new_state[i]

        neighbours.append(
            (calculate_cost(new_state), i, j, new_state)
        )

    return neighbours


def hill_climbing(state):
    step = 0

    print("INITIAL STATE:", state)
    print("INITIAL COST:", calculate_cost(state))
    print("\nINITIAL MATRIX:")
    print_matrix(state)

    while True:
        neighbours = generate_neighbours(state)
        neighbours.sort(key=lambda x: (x[0], x[1], x[2]))

        print("\nALL NEIGHBOURS:")

        for cost, i, j, new_state in neighbours:
            print("Swap (x{}, x{}) State: {} Cost: {}".format(
                i, j, new_state, cost
            ))

        best_cost, i, j, best_state = neighbours[0]
        current_cost = calculate_cost(state)

        step += 1

        print("\nSTEP:", step)
        print("Current State:", state)
        print("Current Cost:", current_cost)
        print("Best Swap: (x{}, x{})".format(i, j))
        print("Neighbour State:", best_state)
        print("Neighbour Cost:", best_cost)

        if best_cost >= current_cost:
            print("\nNo better neighbour found.")
            print("Hill Climbing Algorithm Terminated.")
            break

        state = best_state

        if best_cost == 0:
            print("\nGOAL STATE REACHED!")
            break

    print("\nFINAL STATE:", state)
    print("FINAL COST:", calculate_cost(state))
    print("\nFINAL MATRIX:")
    print_matrix(state)


hill_climbing(state)
```
