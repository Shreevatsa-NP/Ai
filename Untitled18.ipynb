```python

facts = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}

rules = [
    (["Enemy(x, America)"], "Hostile(x)"),
    (["Missile(x)"], "Weapon(x)"),
    (["Missile(x)", "Owns(A, x)"],
     "Sells(Robert, x, A)"),
    (["American(p)", "Weapon(q)",
      "Sells(p, q, r)", "Hostile(r)"],
     "Criminal(p)")
]


def match(pattern, fact):
    p = pattern.replace("(", " ").replace(")", "").replace(",", "").split()

    if p[0] != fact[0] or len(p) != len(fact):
        return None

    substitution = {}

    for x, y in zip(p[1:], fact[1:]):
        if x[0].islower():
            if x in substitution and substitution[x] != y:
                return None
            substitution[x] = y
        elif x != y:
            return None

    return substitution


def substitute(expression, sub):
    for var, value in sub.items():
        expression = expression.replace(var, value)
    return expression


def forward_chaining(facts, rules):
    step = 1

    print("INITIAL FACTS:")
    for fact in sorted(facts):
        print(fact)

    while True:
        new_facts = set()

        for conditions, conclusion in rules:
            substitutions = [({}, [])]

            for condition in conditions:
                updated = []

                for sub, used_facts in substitutions:
                    for fact in facts:
                        pattern = substitute(condition, sub)
                        result = match(pattern, fact)

                        if result is not None:
                            merged = sub.copy()
                            merged.update(result)

                            updated.append(
                                (merged, used_facts + [fact])
                            )

                substitutions = updated

            for sub, used_facts in substitutions:
                result = substitute(conclusion, sub)

                predicate = result.replace(
                    "(", " "
                ).replace(")", "").replace(",", "").split()

                inferred_fact = tuple(predicate)

                if inferred_fact not in facts:
                    new_facts.add(inferred_fact)

                    print("\nSTEP", step)
                    print("Rule Used:")
                    print(" AND ".join(conditions), "=>", conclusion)

                    print("Facts Used:")
                    for fact in used_facts:
                        print(fact)

                    print("New Fact Derived:")
                    print(inferred_fact)

                    step += 1

        if not new_facts:
            break

        facts.update(new_facts)

    return facts


result = forward_chaining(facts, rules)

print("\nGOAL CHECKING")

goal = ("Criminal", "Robert")

if goal in result:
    print("Goal:", goal)
    print("RESULT: Robert is Criminal")
else:
    print("Goal cannot be proved.")
    print("shreevatsa N P")
    print("1BM25CS519")
```
