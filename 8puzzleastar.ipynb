```python
import heapq

INITIAL = (
    (2, 8, 3),
    (1, 6, 4),
    (0, 7, 5)
)
GOAL = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)
COUNT_BLANK = False
def h(state):
    misplaced = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != GOAL[i][j]:
                if state[i][j] == 0 and not COUNT_BLANK:
                    continue
                misplaced += 1
    return misplaced
def neighbors(state):
    blank_row = blank_col = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                blank_row, blank_col = i, j
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]
    for row_change, col_change in moves:
        new_row = blank_row + row_change
        new_col = blank_col + col_change
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            board = [list(row) for row in state]

            board[blank_row][blank_col], board[new_row][new_col] = (
                board[new_row][new_col],
                board[blank_row][blank_col]
            )
            yield tuple(tuple(row) for row in board)
def a_star(start, goal):
    counter = 0
    open_list = [
        (h(start), counter, 0, start)
    ]
    parent = {start: None}
    best_g = {start: 0}
    closed = set()
    while open_list:
        f, _, g, state = heapq.heappop(open_list)
        if state in closed:
            continue
        closed.add(state)
        if state == goal:
            path = []
            while state is not None:
                path.append(state)
                state = parent[state]
            return path[::-1], len(closed)
        for next_state in neighbors(state):
            new_g = g + 1
            if next_state not in best_g or new_g < best_g[next_state]:
                best_g[next_state] = new_g
                parent[next_state] = state
                counter += 1
                new_f = new_g + h(next_state)
                heapq.heappush(
                    open_list,
                    (new_f, counter, new_g, next_state)
                )
    return None, len(closed)
def show(state, g):
    heuristic = h(state)
    total = g + heuristic
    print(f"g(n) = {g}\nh(n) = {heuristic}\nf(n) =g(n)+h(n)\n {total}\n")
    for row in state:
        print(" ".join(str(value) if value else "_" for value in row))
    print()
if __name__ == "__main__":
    path, expanded = a_star(INITIAL, GOAL)
    if path is None:
        print("No solution found.")
    else:
        for step, state in enumerate(path):
            print(f"Step {step}")
            show(state, step)
        print(f"goal state f(n): {len(path) - 1}")
        print("Shreevatsa N P")
        print("1BM25CS519")
```
